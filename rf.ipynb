```python
import pandas as pd

from lib.preprocess import preprocess
from lib.config import ExperimentConfig
from lib.experiments import run_grid_search, extract_model_params

df = pd.read_csv("dsat.csv", dtype={352: "str", 353: "str"})
(
    X_train,
    X_val,
    X_test,
    y_train,
    y_val,
    y_test,
    X_train_full,
    y_train_full,
    features,
) = preprocess(df, feature_set_num=1)
```

```python
base_rf_cfg = ExperimentConfig(
    name="rf_grid",
    model_type="rf",
    params={
        "n_estimators": 200,
        "random_state": 42,
        "n_jobs": -1,
    },
    scaler=None,
)
```

```python
param_grid = {
    "model__max_depth": [2, 4, 8],
    "model__min_samples_split": [1, 4, 8],
    "model__min_samples_leaf": [1, 2, 4],
    "model__max_features": ["sqrt"],
}
```

```python
print("grid_rf...")

grid_rf = run_grid_search(
    base_cfg=base_rf_cfg,
    param_grid=param_grid,
    X_train=X_train,
    y_train=y_train,
    scoring="neg_root_mean_squared_error",
    cv=3,
    n_jobs=1
)

print("Best params (raw):", grid_rf.best_params_)
print("Best CV RMSE:", -grid_rf.best_score_)
```
