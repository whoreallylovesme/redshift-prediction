```python
import pandas as pd

from lib.preprocess import preprocess
from lib.config import ExperimentConfig
from lib.experiments import run_grid_search, extract_model_params

df = pd.read_csv("dsat.csv", dtype={352: "str", 353: "str"})
(
    X_train,
    X_val,
    X_test,
    y_train,
    y_val,
    y_test,
    X_train_full,
    y_train_full,
    features,
) = preprocess(df, feature_set_num=1)
```

```python
gam_base_cfg = ExperimentConfig(
    name="gam_grid",
    model_type="gam",
    params={
        "n_splines": 20,
        "lam": 0.6,
        "max_iter": 1000,
    },
)
```

```python
param_grid_gam = {
    "model__n_splines": [10, 20, 30],
    "model__lam": [0.1, 0.6, 1.0],
}
```

```python
print("grid_gam...")

gam_grid = run_grid_search(
    base_cfg=gam_base_cfg,
    param_grid=param_grid_gam,
    X_train=X_train,
    y_train=y_train,
    scoring="neg_root_mean_squared_error",
    cv=3,
    n_jobs=1
)

print("Best params (raw):", gam_grid.best_params_)
print("Best CV RMSE:", -gam_grid.best_score_)
```
